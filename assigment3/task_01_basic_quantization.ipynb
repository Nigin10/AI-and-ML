# Affine INT8 Quantization and Dequantization using NumPy

## Objective
Implement basic affine INT8 quantization and dequantization using NumPy.

### Quantization Formula
\[
q = \text{round}\left(\frac{x}{\text{scale}}\right) + \text{zero\_point}
\]

### Dequantization Formula
\[
x_{dequantized} = (q - \text{zero\_point}) \times \text{scale}
\]

### INT8 Range
- Minimum: **-128**
- Maximum: **127**

### Input
- Tensor = `[1.0, -0.5, 3.2, -2.8, 0.0]`
- Scale = `0.025`
- Zero Point = `0`

```python
# Import NumPy library
import numpy as np
```

## Step 1: Define Input Tensor and Quantization Parameters

The input tensor contains floating-point values that will be converted to
INT8 using affine quantization.

Parameters:
- Scale = 0.025
- Zero Point = 0

```python
# Input tensor
tensor = np.array([1.0, -0.5, 3.2, -2.8, 0.0], dtype=np.float32)

# Quantization parameters
scale = 0.025
zero_point = 0
```

## Step 2: Quantization Function

The function performs the following steps:

1. Divide each value by the scale.
2. Round to the nearest integer.
3. Add the zero point.
4. Clip values to the INT8 range [-128, 127].
5. Convert the result to `np.int8`.

```python
def quantize_tensor(tensor, scale, zero_point):
    """
    Quantize a floating-point tensor into INT8 format.
    """

    # Divide by scale and round to nearest integer
    quantized = np.round(tensor / scale)

    # Add zero point
    quantized = quantized + zero_point

    # Clip values to INT8 range
    quantized = np.clip(quantized, -128, 127)

    # Convert to INT8 datatype
    return quantized.astype(np.int8)
```

## Step 3: Dequantization Function

The function converts the INT8 values back into floating-point values
using the affine dequantization formula.

Formula:

\[
x = (q - zero\_point) \times scale
\]

```python
def dequantize_tensor(quantized_tensor, scale, zero_point):
    """
    Convert an INT8 tensor back to floating-point values.
    """

    dequantized = (quantized_tensor.astype(np.float32) - zero_point) * scale

    return dequantized
```

## Step 4: Perform Quantization and Dequantization

Apply the quantization function followed by the dequantization function.

```python
# Quantize the input tensor
quantized_tensor = quantize_tensor(tensor, scale, zero_point)

# Dequantize the INT8 tensor
dequantized_tensor = dequantize_tensor(quantized_tensor, scale, zero_point)
```

## Step 5: Compute Quantization Error

Evaluate the accuracy of quantization using:

- Absolute Error
- Mean Absolute Error (MAE)
- Maximum Absolute Error

```python
# Calculate absolute error
absolute_error = np.abs(tensor - dequantized_tensor)

# Mean Absolute Error
mae = np.mean(absolute_error)

# Maximum Absolute Error
max_error = np.max(absolute_error)
```

## Step 6: Display Results

Print the original tensor, quantized tensor,
dequantized tensor and error metrics.

```python
print("Original tensor:")
print(tensor)

print("\nQuantized tensor (INT8):")
print(quantized_tensor)

print("\nDequantized tensor (float):")
print(dequantized_tensor)

print("\nAbsolute error per element:")
print(absolute_error)

print("\nMean Absolute Error (MAE):")
print(mae)

print("\nMaximum Absolute Error:")
print(max_error)
```


## Observation

- Most values are reconstructed exactly after dequantization.
- The value **3.2** exceeds the maximum representable INT8 value after scaling (128), so it is clipped to **127**.
- This clipping introduces a quantization error of approximately **0.025**.
- The Mean Absolute Error (MAE) is very small, indicating good reconstruction accuracy.
